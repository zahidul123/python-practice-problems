# Exercise 1: Array Creation and Modification
# Task: Modify the code to create a 4x4 matrix of random integers between 10 and 50, then set all values greater than 40 to 0.

```python
import numpy as np

array = np.random.randint(10, 50, size=(4, 4))
print("Original Array:", array, sep="\n")
array[array > 40] = 0
print("Filtered Array (elements > 40):", array, sep="\n")
```

# Exercise 2: Statistical Operations
# Task: Modify the code to compute the median and standard deviation instead of the mean, and apply it to a 1D array of 20 random floats between 0 and 10.

```python
array= np.random.randint(1,10,size=20)
print("Array:", array)
median=np.median(array)
std_dev=np.std(array)
print("Median:", median)
print("Standard deviation:", std_dev)
```

# Exercise 3: DataFrame Filtering
# Task: Modify the code to filter rows where age is greater than 25 and city is "New York".

```python
import pandas as pd

data = {
    'name': ['Alice', 'Bob', 'Charlie', 'David'],
    'age': [25, 30, 35, 20],
    'city': ['New York', 'Los Angeles', 'New York' , 'Chicago']
}

df = pd.DataFrame(data)
filtered_data=df[ (df['age']>25) & (df['city']=='New York')]
print("Filtered DataFrame:", filtered_data)
```

# Exercise 4: Grouping and Aggregation
# Task: Modify the code to group by city and compute the maximum age for each city.

```python
data = {
    'name': ['Alice', 'Bob', 'Charlie', 'David','Eva'],
    'age': [25, 30, 35, 20,28],
    'city': ['New York', 'Los Angeles', 'New York' , 'Chicago','Los Angeles']
}

df = pd.DataFrame(data)
grouped_by_city=df.groupby('city')['age'].max()
print("Maximum age by city:", grouped_by_city)
```

# Exercise 5: Boxplot Customization
# Task: Modify the code to create a boxplot of total_bill grouped by day, with points colored by sex.

```python
import seaborn as sns

sns.axes_style()
sns.set_style("whitegrid")
sns.set_style("whitegrid", {"axes.facecolor": ".9"})
sns.color_palette("magma", as_cmap=True)

tips = sns.load_dataset("tips")
print('Columns in tips dataset:', tips.columns)

boxplot = sns.boxplot(data=tips,x='day',y='total_bill',hue='sex')

boxplot.set_title("Boxplot of Total Bill by Day and Sex")
```

# Exercise 6: Correlation Heatmap
# Task: Modify the code to create a heatmap of the correlation matrix for numerical columns in the tips dataset, with annotations.

```python
sns.heatmap(tips.corr(numeric_only=True), annot=True, cmap='coolwarm')
import matplotlib.pyplot as plt
correlation_matrix = tips.select_dtypes(include=['number']).corr()

# Create the heatmap
plt.figure(figsize=(8, 6)) # Adjust figure size as needed
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=.5)
plt.title('Correlation Matrix of Numerical Columns in Tips Dataset')
plt.show()
```

# Exercise 8: Bar Plot with Customization
# Task: Modify the code to create a bar plot showing the average total_bill per day, with bars colored by sex.

```python
data = px.data.tips()
bar_plot = px.bar(data, x='day', y='total_bill', color='sex', barmode='group',
                  title='Average Total Bill per Day by Sex',
                  labels={'total_bill':'Average Total Bill', 'day':'Day of the Week'})
bar_plot.show(renderer="browser")
```
